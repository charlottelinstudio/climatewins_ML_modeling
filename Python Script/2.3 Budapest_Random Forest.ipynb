# 1. Import Libraries & Data

```python
import pandas as pd
import numpy as np
import seaborn as sns
import os
import operator
import matplotlib.pyplot as plt
import tensorflow as tf
from sklearn import datasets  
from sklearn.ensemble import RandomForestClassifier
from numpy import argmax
from sklearn.model_selection import train_test_split
from sklearn import metrics  
from sklearn.tree import plot_tree
from sklearn import tree
from sklearn.metrics import accuracy_score
```

```python
# Create data import path
path = "/Users/charlottelin/Documents/11-2025 ClimateWins Machine Learning Analysis/02 Data/Prepared Data/"

# Load data
climate = pd.read_csv(os.path.join(path, 'unscaled_cleaned_2_2.csv'))
prediction = pd.read_csv(os.path.join(path, 'pleasant_weather_cleaned_2_2.csv'))
```

```python
# Check climate import 

print("Data shape:", climate.shape) 
print("Missing values:", climate.isna().sum().sum()) 
climate.head()
```

```python
# Check prediction import
print("Data shape:", prediction.shape) 
print("Missing values:", prediction.isna().sum().sum()) 
prediction.head()
```

# 2. Data Wrangling

```python
climate.isnull().sum()
```

```python
prediction.isnull().sum()
```

```python
# Get all columns that start with 'BUDAPEST'
budapest_columns = [col for col in climate.columns if col.startswith('BUDAPEST')]
print(budapest_columns)
```

```python
# Filter for columns related to the BUDAPEST station
budapest_columns = [col for col in climate.columns if col.startswith("BUDAPEST_")]

# Check results
print("Number of Budapest feature columns:", len(budapest_columns))

# Create a DataFrame with only Budapest data
budapest_df = climate[budapest_columns].copy()

budapest_df.head()
```

```python
budapest_df.shape
```

```python
# Reduce pleasant weather dataset to BUDAPEST only too
prediction_budapest = prediction['BUDAPEST_pleasant_weather']
prediction_budapest
```

```python
print(prediction_budapest.shape)
print(prediction_budapest.value_counts())
```

# 3. Data Shaping

```python
# Define features (X) and target (y) for Budapest
X_budapest = budapest_df
y_budapest = prediction['BUDAPEST_pleasant_weather'].astype(int)

# Convert to NumPy arrays for scikit-learn
X = X_budapest.to_numpy()
y = y_budapest.to_numpy()

# Check shapes
print("Feature matrix shape (X):", X.shape)  
print("Target vector shape (y):", y.shape)   
print("Class balance:\n", pd.Series(y).value_counts())
```

# 4. Data Splitting

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)
```

```python
print(X_train.shape, y_train.shape)
print(X_test.shape, y_test.shape)
```

# 5. Random Forest Modeling

```python
# Train Random Forest
clf = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

clf.fit(X_train, y_train)
```

```python
# Predict + accuracy
y_pred = clf.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print("Random Forest Accuracy on Test Set:", round(accuracy * 100, 2), "%")
```

```python
# Define class names
class_names = ["Unpleasant Weather", "Pleasant Weather"]
```

```python
# Use Budapest feature names for readability
feature_names = list(X_budapest.columns)
```

```python
# Output directory for figures
output_dir = "/Users/charlottelin/Documents/11-2025 ClimateWins Machine Learning Analysis/04 Analysis/Visualizations"
os.makedirs(output_dir, exist_ok=True)
```

```python
# Plot full Budapest tree
tree_idx = 15

fig = plt.figure(figsize=(80, 40))
plot_tree(
    clf.estimators_[tree_idx],
    filled=True,
    feature_names=list(X_budapest.columns),
    class_names=["Unpleasant Weather", "Pleasant Weather"],
    rounded=True,
    fontsize=12
)

plt.title(f"BUDAPEST — Random Forest Tree #{tree_idx} (full depth)")

file_path = os.path.join(output_dir, f"budapest_rf_tree_{tree_idx}_full.png")
plt.savefig(file_path, bbox_inches="tight", dpi=300)
plt.show()
```

```python
# Plot Budapest tree with max depth = 4

tree_idx = 15

fig = plt.figure(figsize=(30, 12))
plot_tree(
    clf.estimators_[tree_idx],
    filled=True,
    feature_names=list(X_budapest.columns),
    class_names=["Unpleasant Weather", "Pleasant Weather"],
    max_depth=4,
    rounded=True,
    fontsize=10
)

plt.title(f"BUDAPEST — Random Forest Tree #{tree_idx} (max_depth=4)")

file_path = os.path.join(output_dir, f"budapest_rf_tree_{tree_idx}_depth4.png")
plt.savefig(file_path, bbox_inches="tight", dpi=300)
plt.show()
```

# 6. Feature Importance Analysis

```python
# Define feature importance variable
feature_importances = clf.feature_importances_
print("Feature importances shape:", feature_importances.shape)
```

```python
# Define feature names used to train the model
budapest_feature_cols = list(X_budapest.columns)
```

```python
# Clean labels for plotting
budapest_features_clean = [c.replace("BUDAPEST_", "") for c in budapest_feature_cols]
```

```python
# Create a Series of importances
important = pd.Series(feature_importances, index=budapest_features_clean).sort_values(ascending=False)
important
```

```python
# Plot + export feature importance

plt.figure(figsize=(10, 6))
important.plot(kind="bar", color='gold', edgecolor="black")

plt.title("Feature Importance — Budapest Weather Station (All Years)", fontsize=14)
plt.xlabel("Weather Feature", fontsize=12)
plt.ylabel("Importance Score", fontsize=12)
plt.xticks(rotation=45, ha="right")
plt.grid(axis="y", linestyle="--", alpha=0.7)

plt.tight_layout()

file_path = os.path.join(output_dir, "budapest_rf_feature_importance_all_years.png")
plt.savefig(file_path, dpi=300, bbox_inches="tight")
plt.show()
```

### <b>Budapest Random Forest — Key Insights</b>

- The Random Forest model achieved 100% accuracy on the test set, indicating that Budapest’s local weather indicators are highly predictive of pleasant days in this dataset.
- Precipitation is the most influential feature by a wide margin, suggesting that the presence or absence of rainfall is the primary driver of whether a day is perceived as pleasant in Budapest.
- Maximum and mean temperature are the next most important features, highlighting the role of warmth in shaping pleasant weather conditions.
- Global radiation and sunshine contribute meaningfully but less strongly, indicating that light conditions refine predictions once precipitation and temperature are accounted for.
- Pressure and humidity have minimal importance, suggesting they add little incremental predictive value for pleasant-day classification in Budapest.
- For Budapest, ClimateWins could prioritize high-quality precipitation and temperature monitoring over additional investment in pressure or humidity sensors when optimizing pleasant-weather prediction models.
